```python
from newutils.data import load_train_gauge, get_train_gauge_ids
from newutils.plots import TSPlotter
```

```python
gauge_ids = get_train_gauge_ids()
```

```python
df = load_train_gauge(gauge_ids[0]).drop(columns="gauge_id").set_index("date").compute()
df.head()
```

```python
YEARS = [2010, 2011]
df = df.query("date.dt.year in @YEARS")
```

```python
TSPlotter.plot_precipitation(df)
```

```python
TSPlotter.plot_temperatures(df)
```

```python
TSPlotter.plot_daily_runoff(df)
```
