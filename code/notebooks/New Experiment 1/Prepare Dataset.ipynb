```python
import logging
from pathlib import Path

import xgboost as xgb
from dask.distributed import Client, LocalCluster
from Dataset import Dataset
from newutils.other.helpers import init_context

init_context()
```

```python
logging.basicConfig(
    level=logging.INFO, format="%(asctime)s - %(name)s - %(levelname)s - %(message)s"
)
```

```python
cluster = LocalCluster(
    dashboard_address=":8789",
    asynchronous=False,
    n_workers=3,
    threads_per_worker=1,
    memory_limit="5GB",
)
client = Client(cluster)
# client = Client("localhost:8786")
```

```python
dataset = Dataset(client)
dataset = dataset.create(536)
```

```python
OUTDIR = Path("artifacts") / "dataset"
OUTDIR.mkdir(parents=True, exist_ok=True)
dataset.df.compute().to_parquet(Path("artifacts") / "dataset" / "full_parquet.parquet")
```
